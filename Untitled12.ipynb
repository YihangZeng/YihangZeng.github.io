```python
##ori
import h5py
import scipy.sparse
import time
import random
import numpy as np
import pandas as pd
import torch
import scipy.sparse as sp
from sklearn.metrics.cluster import adjusted_rand_score
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
import torch.nn.functional as F

class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels)
        self.conv3 = GCNConv(hidden_channels, hidden_channels)
        self.conv4 = GCNConv(hidden_channels, out_channels)

    def forward(self, x, edge_index):
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv1(x, edge_index).relu()
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv2(x, edge_index).relu()
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv3(x, edge_index).relu()
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv4(x, edge_index)
        return x


def preprocess_labels(y_label, num_classes, alpha=0.1, beta=0.9):
    mask = np.random.rand(len(y_label)) > (1 - alpha)  # 遮蔽 alpha 百分比标签
    y_new = y_label.copy()
    y_new[~mask] = -1  # 未标记标签设为 -1

    for i in range(len(y_new)):
        if y_new[i] != -1 and random.random() < beta:
            y_new[i] = random.randint(0, num_classes - 1)  # 随机引入错误标签

    return y_new  # 返回 1D 索引形式

df = pd.DataFrame(columns=['MR', 'ARI_o', 'ARI_r', 'Time'])

start = time.time()

with h5py.File('dataset1.h5ad', 'r') as f:
    group = f['obsp']['connectivities']

    data = group['data'][:]
    indices = group['indices'][:]
    indptr = group['indptr'][:]
    shape = (
        f['obsp']['connectivities'].attrs['shape'][0],
        f['obsp']['connectivities'].attrs['shape'][1]
    )

    mat = scipy.sparse.csr_matrix((data, indices, indptr), shape=shape)

coo = mat.tocoo()
rows, cols, data = coo.row, coo.col, coo.data

with h5py.File('dataset1.h5ad', 'r') as h5file:
    obs_group = h5file['obs']
    if "codes" in obs_group['annotation']:
      mat = obs_group['annotation']['codes'][:]
    else:
      mat = obs_group['annotation'][:]

val = {}
for i in mat:
    if i not in val:
        val[i] = len(val)

num_classes = len(val)
# print(coo)
# print(mat)

epochs = 200

k1 = 1.00

edge_index = torch.tensor(np.vstack((rows, cols)), dtype=torch.long)
x = torch.eye(mat.shape[0], dtype=torch.float32)
y_label = torch.tensor([val[m] for m in mat], dtype=torch.long)
num_nodes = y_label.size(0)

random_indices = random.sample(range(num_nodes), int(num_nodes * 0.1))
selected_mask = torch.zeros(num_nodes, dtype=torch.bool)
selected_mask[random_indices] = True
y_label_selected = torch.full_like(y_label, -1)
y_label_selected[selected_mask] = y_label[selected_mask]
print(y_label_selected)
data = Data(x=x, edge_index=edge_index, y=y_label)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = GCN(in_channels=x.size(1), hidden_channels=32, out_channels=num_classes).to(device)
data = data.to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

start_time = time.perf_counter()
# 训练 GCN
model.train()
for epoch in range(epochs):
    optimizer.zero_grad()
    out = model(data.x, data.edge_index)
    loss = F.cross_entropy(out, data.y, ignore_index=-1)
    loss.backward()
    optimizer.step()



while True:
    for _ in range(10):
        # 构造 PyTorch Geometric 数据对象
        y_new = preprocess_labels(y_label.numpy(), num_classes, alpha=k1, beta=(1 - k1))
        y_new[selected_mask] = y_label[selected_mask]
        # y_new = y_new.argmax(axis=1)

        # 测试 GCN
        model.eval()
        out = model(data.x, data.edge_index).argmax(dim=1).cpu().numpy()
        out[selected_mask] = y_label[selected_mask]
        ari_o = adjusted_rand_score(y_new, out)

        y_res = out
        # y_res = out.copy()
        # y_res[selected_mask] = y_label[selected_mask]
        ari_r = adjusted_rand_score(y_new, y_res)

        end_time = time.perf_counter()
        execution_time = end_time - start_time

        item = [round((1 - k1), 2), ari_o, ari_r, round(execution_time, 5)]
        print(item)
        df.loc[len(df)] = item

    k1 -= 0.05
    print(k1)
    # break

    if k1 < 0:
        break

df.to_csv("mistake.tsv", sep='\t', header=True, index=True)
end = time.time()
print(f"time :{end - start}")
```
